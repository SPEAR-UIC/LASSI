# Manage LLM: Download, Unload from memory, Delete 

```python
# List the available models installed on this Ollama server
# display(ollama.list().model_dump())
models_loaded = ollama.list().model_dump()
num = 1
for model in models_loaded["models"]:
    print(f"{num}. {model['model']}")
    num += 1
```

```python
# DOWNLOAD a new model. Download of weights will happen if model does not exist locally.
llm_model = "gpt-oss:20b"  # "deepseek-r1:70b"  granite-code:34b-instruct-q8_0
ollama.pull(llm_model)
```

```python
# DELETE a model from this server
llm_model = "deepseek-coder-v2:16b"
ollama.delete(llm_model)
```

```python
# UNLOAD a model from memory without deleting the LLM
llm_model = "llama3.3:70b"
ollamastream.unload_ollama_model(llm_model)
```

# TEST Ollama served LLMs

```python
llm = "gpt-oss:120b"
systemprompt = "You are an expert coder."
userprompt = "How do you print hello world in Python?"
temperature = 0.2
verifyprompt = True

start_time = time.time() 
response = ollamastream.ollama_llm(llm, systemprompt, userprompt, temperature, verifyprompt)
end_time = time.time() - start_time
print("\nLLM response time: " + str(round(end_time, 2)) + " seconds.")
```
